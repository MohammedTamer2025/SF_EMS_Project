## 5. Null Analysis & Handling

```python
nulls = df.isnull().sum()
null_pct = (nulls / len(df)) * 100

null_report = pd.DataFrame({
    'null_count': nulls,
    'null_pct': null_pct
}).query('null_count > 0').sort_values('null_pct', ascending=False)

print('Columns with nulls:')
print(null_report.to_string())
print(f'\nTotal rows: {len(df):,}')
```

### 5.1 Null Handling Strategy

Three kinds of nulls:

1. **Truly missing** -> imputed
2. **Not applicable** -> kept as NaN (a value factually does not exist)
3. **Unknown outcome** -> kept as NaN (cannot be invented)

| Column | Null % | Kind | Handling |
|--------|--------|------|----------|
| `transport_dttm` / `hospital_dttm` / `transport_time_min` | ~76% | Not applicable (no patient transport) | Keep NaN |
| `on_scene_dttm` / `response_time_min` | 21.7% | Unknown outcome (cancelled / no on-scene) | Keep NaN; ML drops these only when predicting response time |
| `response_dttm` | 2.7% | Missing | Keep NaN |
| `holiday_name` | 96% | Not applicable (not a holiday) | Keep NaN; use `is_holiday` flag |
| `call_type_group` | 2.2% | Missing | Fill `'Unknown'` |
| `original_priority` | 0.9% | Missing | Fill from `priority` |
| `fire_prevention_district` | 0.6% | Missing | Fill mode |
| `zipcode_of_incident` | 0.1% | Missing | Fill mode |
| `city` | 0.1% | Missing | Fill `'San Francisco'` |
| `neighborhoods_analysis_boundaries` | ~0% | Missing | Fill `'Unknown'` |
| `supervisor_district` | ~0% | Missing | Fill mode |
| `station_area` / `box` | ~0% | Missing | Fill mode |

**Why NOT impute the time targets?**
- `response_time_min` is the ML outcome: imputing it would fabricate results the model would then "predict". Rows without an on-scene time are dropped only inside the ML step.
- `transport_time_min` cleanly does not exist for the ~76% of calls without a patient transport.

```python
# --- Impute only the truly-missing values ---
df['call_type_group'] = df['call_type_group'].fillna('Unknown')
df['neighborhoods_analysis_boundaries'] = df['neighborhoods_analysis_boundaries'].fillna('Unknown')

df['original_priority'] = df['original_priority'].fillna(df['priority'])
df['city'] = df['city'].fillna('San Francisco')

for col in ['zipcode_of_incident', 'fire_prevention_district', 'supervisor_district', 'station_area', 'box']:
    mode_val = df[col].mode(dropna=True)
    if not mode_val.empty:
        df[col] = df[col].fillna(mode_val.iloc[0])

print('Remaining nulls after imputation:')
print(df[['city', 'original_priority', 'call_type_group', 'zipcode_of_incident',
          'neighborhoods_analysis_boundaries', 'fire_prevention_district',
          'supervisor_district', 'station_area', 'box']].isnull().sum())
print('\nStill-NaN (kept on purpose):')
still = ['response_dttm', 'on_scene_dttm', 'transport_dttm', 'hospital_dttm', 'holiday_name']
print(df[still].isnull().sum())
```

## 7. Feature Engineering

Create derived columns needed for EDA and modeling.

```python
df['response_time_min'] = (df['on_scene_dttm'] - df['received_dttm']).dt.total_seconds() / 60
df['dispatch_lag_min'] = (df['dispatch_dttm'] - df['entry_dttm']).dt.total_seconds() / 60
df['transport_time_min'] = (df['hospital_dttm'] - df['transport_dttm']).dt.total_seconds() / 60

print('response_time_min:')
print(f'  Non-null: {df["response_time_min"].notna().sum():,}')
print(f'  Null: {df["response_time_min"].isna().sum():,}')
print(f'  Mean: {df["response_time_min"].mean():.2f} min')
print(f'  Median: {df["response_time_min"].median():.2f} min')

print('\ndispatch_lag_min:')
print(f'  Non-null: {df["dispatch_lag_min"].notna().sum():,}')
print(f'  Mean: {df["dispatch_lag_min"].mean():.2f} min')
```

```python
df['hour_of_day'] = df['received_dttm'].dt.hour
df['day_of_week'] = df['received_dttm'].dt.day_name()
df['month'] = df['received_dttm'].dt.month
df['year'] = df['received_dttm'].dt.year
df['is_weekend'] = df['day_of_week'].isin(['Saturday', 'Sunday'])
df['is_als'] = df['als_unit'].astype(int) #ALS = Advanced Life Support 

print('New time columns created:')
print(df[['received_dttm', 'hour_of_day', 'day_of_week', 'month', 'is_weekend', 'is_als']].head())
```

## 7b. IQR Outlier Detection (Flag-Only)

Standard Tukey fences (1.5x IQR) on the derived time columns.
- **Flag**: `is_outlier_*` columns added (row = True if outside the fence)
- **Values are NOT changed** - the CSV keeps the real measured times
- **Negative times** (data inconsistencies) fall below the lower fence (floored at 0) -> automatically flagged

Why flag-only instead of capping?
- Capping **crushes model performance**: `R2 = 0.56` on capped response times vs `R2 = 0.80` on the real values.
- Keep the real values; filter with the flags in Excel / Power BI / ML when desired (`is_outlier_* == False`).

```python
iqr_cols = ['response_time_min', 'dispatch_lag_min', 'transport_time_min']

for col in iqr_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = max(0, Q1 - 1.5 * IQR)
    upper = Q3 + 1.5 * IQR

    is_outlier = (df[col] < lower) | (df[col] > upper)
    df[f'is_outlier_{col}'] = is_outlier

    outlier_count = is_outlier.sum()
    pct = outlier_count / len(df) * 100
    print(f'{col}:')
    print(f'  Q1={Q1:.2f}, Q3={Q3:.2f}, IQR={IQR:.2f}')
    print(f'  Fences: [{lower:.2f}, {upper:.2f}]')
    print(f'  Outliers flagged: {outlier_count:,} ({pct:.2f}%)')
    print(f'  Values unchanged: min={df[col].min():.2f}, max={df[col].max():.2f}')

print('\nFlag-only complete: is_outlier_* columns added, values NOT modified.')
```
