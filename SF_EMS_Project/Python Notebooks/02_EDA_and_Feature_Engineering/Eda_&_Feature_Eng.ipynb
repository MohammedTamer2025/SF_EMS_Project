```python
#create Time Period feature
def get_time_period(hour):
    if 0 <= hour < 6:
        return 'Night'
    elif 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 18:
        return 'Afternoon'
    else:
        return 'Evening'
df['time_period'] = df['hour_of_day'].apply(get_time_period)
```

```python
df['time_period'].value_counts()
```

```python
#Response Time Category
def categorize_response_time(time):
    if pd.isna(time):
        return 'Unknown'
    elif time <= 5:
        return 'Fast'
    elif time <= 10:
        return 'Moderate'
    else:
        return 'Slow'

df['response_time_category'] = df['response_time_min'].apply(
    categorize_response_time
)
```

```python
#Dispatch Lag Category
def categorize_dispatch_lag(time):
    if pd.isna(time):
        return 'Unknown'
    elif time <= 2:
        return 'Fast Dispatch'
    elif time <= 5:
        return 'Moderate Dispatch'
    else:
        return 'Delayed Dispatch'
df['dispatch_lag_category'] = df['dispatch_lag_min'].apply(
    categorize_dispatch_lag
)
```

```python
#Units per Incident
df['units_per_incident'] = (
    df.groupby('incident_number')['unit_id']
      .transform('nunique')
)
```
